## 12. Potential Target and Leakage Concerns

Several columns require investigation as potential target variables, proxies, post-outcome variables, or identifiers that could lead to data leakage:

*   `label_functional_status_clean`: This column directly indicates the water point's functional status (e.g., 0.0 or 1.0) and is a strong candidate for a primary target variable. Its ID counterpart is `label_functional_status_id`.
*   `status_clean`: Similar to `label_functional_status_clean`, this provides detailed status information (e.g., 'Non-Functional', 'Functional', 'Functional, with issues'). This is likely a more granular version of the target or a direct proxy.
*   `status_id`: An identifier for the `status_clean` column, also a direct proxy.
*   `wpdx_crucialness_score_LEAKY`, `wpdx_pressure_score_LEAKY`, `wpdx_rehab_priority_LEAKY`, `wpdx_pop_would_gain_access_LEAKY`: These columns have the suffix `_LEAKY`, which strongly suggests they are derived from or directly related to the outcome (e.g., functionality or maintenance need) and could lead to data leakage if used as features directly.
*   `master_id`, `wpdx_id`, `row_id_export`: These are unique identifiers for each water point/record and should be excluded from feature sets to prevent data leakage and ensure generalizability.
*   `wpdx_is_duplicate_flag`: This flag indicates if a record is a duplicate, which might be a post-processing artifact or a derived feature. Using it directly might introduce leakage.
*   `dist_nearest_functional_other_m_LEAKY`: This distance to *functional* water points may be correlated with or directly influenced by the target and needs careful review for leakage.

## Build Preprocessing Pipeline

Create a preprocessing pipeline for numerical, categorical, and boolean features.


````python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
````

````python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

boolean_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features),
        ("boolean", boolean_pipeline, boolean_features)
    ]
)

print("Preprocessing pipeline created successfully.")
````

## Fix Boolean Preprocessing

Convert boolean features to numeric values before preprocessing.


````python
from sklearn.preprocessing import FunctionTransformer

boolean_pipeline = Pipeline([
    ("to_numeric", FunctionTransformer(
        lambda x: x.astype(int),
        feature_names_out="one-to-one"
    )),
    ("imputer", SimpleImputer(strategy="most_frequent"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features),
        ("boolean", boolean_pipeline, boolean_features)
    ]
)

print("Boolean preprocessing updated successfully.")
````

````python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

results = []

for name, model in trained_models.items():

    # Predictions
    y_pred = model.predict(X_test_processed)
    y_prob = model.predict_proba(X_test_processed)[:, 1]

    # Metrics for Functional class (1)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    # Overall ranking metrics
    roc_auc = roc_auc_score(y_test, y_prob)
    pr_auc = average_precision_score(y_test, y_prob)

    accuracy = accuracy_score(y_test, y_pred)

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "PR-AUC": pr_auc,
        "ROC-AUC": roc_auc
    })

results_df = pd.DataFrame(results)

results_df.sort_values(
    by="F1",
    ascending=False
).reset_index(drop=True)
````

## Risk-Focused Evaluation

````python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    average_precision_score
)

risk_results = []

for name, model in trained_models.items():

    # Predictions
    y_pred = model.predict(X_test_processed)

    # Probability of Non-Functional class (class 0)
    y_prob_nonfunctional = model.predict_proba(
        X_test_processed
    )[:, 0]

    # Treat class 0 as the positive class
    precision = precision_score(
        y_test,
        y_pred,
        pos_label=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        pos_label=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        pos_label=0
    )

    # Convert target: 0 = Non-Functional → 1
    y_test_risk = (y_test == 0).astype(int)

    pr_auc = average_precision_score(
        y_test_risk,
        y_prob_nonfunctional
    )

    risk_results.append({
        "Model": name,
        "Non-Functional Precision": precision,
        "Non-Functional Recall": recall,
        "Non-Functional F1": f1,
        "Non-Functional PR-AUC": pr_auc
    })

risk_results_df = pd.DataFrame(risk_results)

risk_results_df.sort_values(
    by="Non-Functional F1",
    ascending=False
).reset_index(drop=True)
````

## Spatial Cross-Validation of Random Forest

The Random Forest model is evaluated across geographically separated regions while fitting preprocessing separately within each fold.


````python
from sklearn.model_selection import GroupKFold
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    FunctionTransformer
)
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

group_kfold = GroupKFold(n_splits=5)

spatial_results = []

for fold, (train_idx, val_idx) in enumerate(
    group_kfold.split(
        X_spatial,
        y_spatial,
        groups=groups_spatial
    ),
    start=1
):

    X_train_fold = X_spatial.iloc[train_idx]
    X_val_fold = X_spatial.iloc[val_idx]

    y_train_fold = y_spatial.iloc[train_idx]
    y_val_fold = y_spatial.iloc[val_idx]

    # Identify feature types
    numeric_fold = X_train_fold.select_dtypes(
        include=["number"]
    ).columns.tolist()

    categorical_fold = X_train_fold.select_dtypes(
        include=["object", "string"]
    ).columns.tolist()

    boolean_fold = X_train_fold.select_dtypes(
        include=["bool"]
    ).columns.tolist()

    # Preprocessing
    numeric_pipeline_fold = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    categorical_pipeline_fold = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ])

    boolean_pipeline_fold = Pipeline([
        ("to_numeric", FunctionTransformer(
            lambda x: x.astype(int),
            feature_names_out="one-to-one"
        )),
        ("imputer", SimpleImputer(strategy="most_frequent"))
    ])

    preprocessor_fold = ColumnTransformer(
        transformers=[
            ("numeric", numeric_pipeline_fold, numeric_fold),
            ("categorical", categorical_pipeline_fold, categorical_fold),
            ("boolean", boolean_pipeline_fold, boolean_fold)
        ]
    )

    # Fit preprocessing ONLY on training regions
    X_train_processed_fold = preprocessor_fold.fit_transform(
        X_train_fold
    )

    X_val_processed_fold = preprocessor_fold.transform(
        X_val_fold
    )

    # Train Random Forest
    rf_fold = RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    )

    rf_fold.fit(
        X_train_processed_fold,
        y_train_fold
    )

    # Predictions
    y_pred_fold = rf_fold.predict(
        X_val_processed_fold
    )

    y_prob_nonfunctional = rf_fold.predict_proba(
        X_val_processed_fold
    )[:, 0]

    # Convert risk class:
    # 0 = Non-Functional → 1
    y_val_risk = (y_val_fold == 0).astype(int)
    y_pred_risk = (y_pred_fold == 0).astype(int)

    spatial_results.append({
        "Fold": fold,
        "Precision": precision_score(
            y_val_risk,
            y_pred_risk
        ),
        "Recall": recall_score(
            y_val_risk,
            y_pred_risk
        ),
        "F1": f1_score(
            y_val_risk,
            y_pred_risk
        ),
        "PR-AUC": average_precision_score(
            y_val_risk,
            y_prob_nonfunctional
        ),
        "ROC-AUC": roc_auc_score(
            y_val_fold,
            rf_fold.predict_proba(X_val_processed_fold)[:, 1]
        )
    })

spatial_results_df = pd.DataFrame(spatial_results)

spatial_results_df
````

## 11. Feature Engineering

Feature engineering is the process of creating new predictive variables from
existing data to help the machine learning model capture meaningful patterns
that may not be represented directly by the original features.

Based on the exploratory data analysis, false-negative analysis, Random Forest
feature importance, and permutation importance, the main patterns identified
in the dataset are related to:

- Water-point age and infrastructure lifetime
- Physical accessibility and geographic isolation
- Population pressure around water points
- Local water-point density
- Recent rainfall and deviation from normal rainfall
- Seasonal timing of the survey

Rather than creating a large number of new variables, we will create a small
set of domain-informed features. Each engineered feature should have a clear
interpretation and a plausible relationship with water-point functionality.

The engineered features will then be evaluated against the existing baseline
model to determine whether they improve predictive performance without
introducing unnecessary complexity or data leakage.

### 11.1 Water-Point Age Features

The age of a water point is one of the strongest predictors identified by the
baseline Random Forest and permutation importance analysis.

To allow the model to capture broad infrastructure life stages, we create
age-based categories in addition to the continuous `calculated_age` variable.

These categories are intended to represent relatively young, established,
and older water points. The original continuous age variable is retained so
that the model can still learn finer-grained age effects.

````python
# Create broad water-point age groups
df_clean["age_group"] = pd.cut(
    df_clean["calculated_age"],
    bins=[-1, 5, 10, 20, float("inf")],
    labels=["0-5", "6-10", "11-20", "20+"]
)

# Check distribution
df_clean["age_group"].value_counts(dropna=False)
````

### 11.2 Accessibility and Isolation Features

The false-negative analysis showed differences in several distance-based
variables, including distance to cities, roads, health facilities, and other
water points.

These variables describe different aspects of physical accessibility.
Therefore, we create summary indicators that represent the overall
accessibility burden of a water point.

The purpose is to provide the model with a more compact representation of
potential geographic isolation while retaining the original distance
variables for comparison.

All transformations use information available at or before the survey
observation and do not use the target variable.

````python
# Distances that represent physical accessibility
accessibility_features = [
    "wpdx_dist_city_m",
    "wpdx_dist_primary_road_m",
    "wpdx_dist_secondary_road_m",
    "wpdx_dist_tertiary_road_m",
    "dist_nearest_health_facility_m",
    "dist_nearest_school_m"
]

# Check that all expected columns exist
missing_accessibility = [
    col for col in accessibility_features
    if col not in df_clean.columns
]

print("Missing accessibility features:", missing_accessibility)
````

````python
# Create a combined accessibility burden using standardized distances
from sklearn.preprocessing import StandardScaler

accessibility_values = df_clean[accessibility_features].copy()

accessibility_scaled = StandardScaler().fit_transform(
    accessibility_values
)

df_clean["accessibility_burden"] = accessibility_scaled.mean(axis=1)

df_clean["accessibility_burden"].describe()
````

### 11.3 Local Water-Point Density

The dataset contains the number of nearby water points at different spatial
radii.

These variables provide information about the local availability of alternative
water sources. We create a density-oriented feature using the number of water
points within 1 km.

This feature is intended to capture local water-point concentration while
avoiding the creation of many highly related variables.

````python
df_clean["water_points_density_1km"] = (
    df_clean["n_water_points_within_1000m"] / 1.0
)

df_clean["water_points_density_1km"].describe()
````

````python
import numpy as np

df_clean["water_points_density_1km2"] = (
    df_clean["n_water_points_within_1000m"] /
    (np.pi * 1**2)
)
````

### 11.4 Rainfall Stress Features

Rainfall variables describe both the amount of recent rainfall and how current
rainfall compares with the long-term normal.

Relative-to-normal variables are particularly useful because the same amount of
rainfall can represent very different conditions in different climatic regions.

We therefore create rainfall stress indicators from the existing
percentage-of-normal variables.

A value below 100% indicates rainfall below the long-term normal, while values
above 100% indicate rainfall above normal conditions.

````python
# Rainfall deficit relative to normal
df_clean["rain_3m_deficit_pct"] = (
    100 - df_clean["rain_3m_pct_of_normal"]
)

df_clean["rain_6m_deficit_pct"] = (
    100 - df_clean["rain_6m_pct_of_normal"]
)

df_clean["rain_12m_deficit_pct"] = (
    100 - df_clean["rain_12m_pct_of_normal"]
)

# Positive values = below normal rainfall
# Negative values = above normal rainfall
````

````python
df_clean["rainfall_stress_12m"] = (
    df_clean["rain_12m_pct_of_normal"] < 80
).astype(int)

df_clean["rainfall_stress_6m"] = (
    df_clean["rain_6m_pct_of_normal"] < 80
).astype(int)
````

## 13. Probability Threshold Analysis

The Random Forest model produces a probability for each class. Instead of
using the default classification threshold only, different probability
thresholds are evaluated for the Non-Functional class.

For this analysis:

- Class 0 = Non-Functional
- Class 1 = Functional

A water point will be classified as Non-Functional when its predicted
probability of being Non-Functional is greater than or equal to the selected
threshold.

The purpose of this analysis is to understand the trade-off between:

- Precision
- Recall
- F1-score
- False Negatives
- False Positives

This is important because false negatives represent water points that are
actually Non-Functional but are predicted as Functional.

````python
import numpy as np
import pandas as pd

# Use the final Random Forest model
rf_model = trained_models["Random Forest"]

# Check class order
print("Model classes:", rf_model.classes_)

# Find the column corresponding to class 0 (Non-Functional)
nonfunctional_class_index = list(rf_model.classes_).index(0)

# Probability that each water point is Non-Functional
proba_nonfunctional = rf_model.predict_proba(
    X_test_processed
)[:, nonfunctional_class_index]

print("Number of probability predictions:", len(proba_nonfunctional))
print("Minimum probability:", proba_nonfunctional.min())
print("Maximum probability:", proba_nonfunctional.max())
````

### 13.1 Evaluate Different Probability Thresholds

Several probability thresholds will be tested.

A lower threshold makes the model more likely to classify a water point as
Non-Functional. This can increase recall and reduce false negatives, but may
also increase false positives.

A higher threshold has the opposite effect.

The thresholds will therefore be compared before selecting the threshold
for the deployment model.

````python
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# Convert the actual target into:
# 1 = Non-Functional
# 0 = Functional
y_test_nonfunctional = (y_test == 0).astype(int)

# Thresholds to evaluate
thresholds = np.arange(0.30, 0.71, 0.05)

threshold_results = []

for threshold in thresholds:

    # Predict Non-Functional when probability >= threshold
    pred_nonfunctional = (
        proba_nonfunctional >= threshold
    ).astype(int)

    # Confusion matrix:
    # TN = correctly identified Functional
    # FP = Functional predicted as Non-Functional
    # FN = Non-Functional predicted as Functional
    # TP = correctly identified Non-Functional
    tn, fp, fn, tp = confusion_matrix(
        y_test_nonfunctional,
        pred_nonfunctional,
        labels=[0, 1]
    ).ravel()

    threshold_results.append({
        "threshold": round(float(threshold), 2),

        "precision_non_functional": precision_score(
            y_test_nonfunctional,
            pred_nonfunctional,
            zero_division=0
        ),

        "recall_non_functional": recall_score(
            y_test_nonfunctional,
            pred_nonfunctional,
            zero_division=0
        ),

        "f1_non_functional": f1_score(
            y_test_nonfunctional,
            pred_nonfunctional,
            zero_division=0
        ),

        "false_negatives": int(fn),
        "false_positives": int(fp)
    })

threshold_results_df = pd.DataFrame(threshold_results)

threshold_results_df
````

### 13.2 Threshold Trade-off

The threshold results show how changing the probability threshold affects
the detection of Non-Functional water points.

Lower thresholds generally make the model more sensitive to potential
Non-Functional water points. This can reduce false negatives but may
increase false positives.

Higher thresholds generally make the model more conservative when
classifying a water point as Non-Functional.

The threshold will therefore be selected based on the operational objective

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.plot(
    threshold_results_df["threshold"],
    threshold_results_df["precision_non_functional"],
    marker="o",
    label="Precision"
)

plt.plot(
    threshold_results_df["threshold"],
    threshold_results_df["recall_non_functional"],
    marker="o",
    label="Recall"
)

plt.plot(
    threshold_results_df["threshold"],
    threshold_results_df["f1_non_functional"],
    marker="o",
    label="F1-score"
)

plt.xlabel("Probability Threshold")
plt.ylabel("Score")
plt.title("Random Forest Threshold Trade-off")
plt.xticks(threshold_results_df["threshold"])
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

### 13.3 Threshold Selection

Based on the threshold analysis, a probability threshold of 0.40 is selected
as the initial deployment threshold for identifying Non-Functional water
points.

At this threshold:

- Precision for Non-Functional = 73.87%
- Recall for Non-Functional = 84.08%
- F1-score for Non-Functional = 78.65%
- False Negatives = 242
- False Positives = 452

The selected threshold prioritizes the detection of Non-Functional water
points while maintaining a reasonable balance with false-positive
predictions.

The threshold is applied to the model probability output and does not
require retraining the Random Forest model.

````python
# Final deployment threshold
FINAL_THRESHOLD = 0.40

print("Selected deployment threshold:", FINAL_THRESHOLD)
````

### 13.4 Threshold Analysis Conclusion

The threshold analysis demonstrated that changing the probability threshold
changes the balance between detecting Non-Functional water points and
generating false-positive alerts.

A threshold of 0.40 will therefore be carried forward for the final
evaluation and deployment pipeline.

No additional model training or feature engineering will be performed.
